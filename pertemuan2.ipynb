```python
# Membaca citra dan konversi ke RGB
img = cv.imread('frieren-beyond-5120x2880-22999.jpg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Melakukan 3 bentuk pencerminan (flipping)
flip_vertikal = cv.flip(img_rgb, 0)   # flipCode 0: vertikal (atas-bawah)
flip_horizontal = cv.flip(img_rgb, 1) # flipCode 1: horizontal (kiri-kanan)
flip_kedua = cv.flip(img_rgb, -1)     # flipCode -1: vertikal dan horizontal sekaligus

# Menampilkan dalam satu figure berisi 3 subplot dengan label
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertikal)
plt.title('Pencerminan Vertikal')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title('Pencerminan Horizontal')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_kedua)
plt.title('Pencerminan Keduanya')
plt.axis('off')

plt.show()
```

```python
# Load image and convert to RGB
img = cv.imread('frieren-beyond-5120x2880-22999.jpg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# --- Task 1: Red-Blue & Green-Blue Channel Combinations ---
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0  # Zero out Green channel

img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0  # Zero out Red channel

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title('Red-Blue Combination (Green Zeroed)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title('Green-Blue Combination (Red Zeroed)')
plt.axis('off')
plt.show()

# --- Task 2: Three Types of Flipping (Vertical, Horizontal, Both) ---
flip_v = cv.flip(img_rgb, 0)   # Vertical
flip_h = cv.flip(img_rgb, 1)   # Horizontal
flip_b = cv.flip(img_rgb, -1)  # Both

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(flip_v)
plt.title('Vertical Flip')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_h)
plt.title('Horizontal Flip')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_b)
plt.title('Both Flips')
plt.axis('off')
plt.show()

# --- Task 3: Draw Rectangle, Circle, and Name Text ---
img_annotated = img_rgb.copy()
cv.rectangle(img_annotated, pt1=(160, 40), pt2=(220, 150), color=(255, 0, 0), thickness=3)
cv.circle(img_annotated, center=(190, 90), radius=35, color=(0, 255, 0), thickness=2)
cv.putText(img_annotated, text="Angelina", org=(100, 200), fontFace=cv.FONT_HERSHEY_SIMPLEX,
           fontScale=1, color=(255, 255, 0), thickness=2, lineType=cv.LINE_AA)

plt.figure(figsize=(7, 7))
plt.imshow(img_annotated)
plt.title('Annotated Image with Shapes and Name')
plt.axis('off')
plt.show()
```

```python
img = cv.imread('frieren-beyond-5120x2880-22999.jpg')
h, w = img.shape[:2]
original_mean = np.mean(img)

# Shrink down and restore back to original dimensions
small = cv.resize(img, (w // 4, h // 4), interpolation=cv.INTER_LINEAR)
restored = cv.resize(small, (w, h), interpolation=cv.INTER_LINEAR)
restored_mean = np.mean(restored)

print(f"Original Mean Intensity: {original_mean}")
print(f"Restored Mean Intensity: {restored_mean}")
```

```python
def add_adaptive_label(image_path, label_text):
    img = cv.imread(image_path)
    h, w = img.shape[:2]

    # Dynamically calculate bar height, font scale, and thickness based on image resolution
    bar_height = int(h * 0.12)
    font_scale = max(0.6, w / 1000.0)
    thickness = max(1, int(w / 400))

    # Draw a solid black background bar at the bottom
    cv.rectangle(img, (0, h - bar_height), (w, h), (0, 0, 0), -1)

    # Position text proportionally inside the bar
    text_org = (int(w * 0.03), h - int(bar_height * 0.35))
    cv.putText(img, label_text, text_org, cv.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Applying the same adaptive function to an image
processed_img = add_adaptive_label('frieren-beyond-5120x2880-22999.jpg', 'Documentation Event 2026')

plt.figure(figsize=(10, 5))
plt.imshow(processed_img)
plt.axis('off')
plt.show()
```

```python
def create_square_thumbnail(image_path, store_name):
    img = cv.imread(image_path)
    h, w = img.shape[:2]

    # Square dimension based on the larger side to maintain aspect ratio
    max_dim = max(h, w)

    # Create a white canvas (1:1 square)
    canvas = np.ones((max_dim, max_dim, 3), dtype=np.uint8) * 255

    # Compute coordinates to center the image
    y_offset = (max_dim - h) // 2
    x_offset = (max_dim - w) // 2

    # Place image in the center of the canvas
    canvas[y_offset:y_offset+h, x_offset:x_offset+w] = img

    # Adaptive scaling for store name text/logo
    font_scale = max(0.5, max_dim / 1000.0)
    thickness = max(1, int(max_dim / 500))

    # Add store name text proportionally near the bottom
    text_org = (int(max_dim * 0.05), int(max_dim * 0.95))
    cv.putText(canvas, store_name, text_org, cv.FONT_HERSHEY_SIMPLEX, font_scale, (0, 0, 0), thickness, cv.LINE_AA)

    return cv.cvtColor(canvas, cv.COLOR_BGR2RGB)

# Apply the function to create a 1:1 square thumbnail with a store label
thumbnail = create_square_thumbnail('frieren-beyond-5120x2880-22999.jpg', 'Official Store - 2026')

plt.figure(figsize=(7, 7))
plt.imshow(thumbnail)
plt.axis('off')
plt.show()
```
